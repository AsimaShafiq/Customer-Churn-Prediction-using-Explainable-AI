## Importing Required Libraries:



````python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer                        # performs  different transformations on different columns
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.pipeline import Pipeline                                # pipeline -> sequence of steps
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier                                    # Gradient boosting classifier


import warnings
warnings.filterwarnings("ignore")
````

## Load Final Dataset

The final feature-engineered dataset created in Notebook 04 is loaded.

No additional data cleaning or feature removal is performed in this notebook because those tasks were completed in the previous phases.

````python
file_path = "/content/Featured_telco_churn.csv"
df = pd.read_csv(file_path,encoding = 'ISO-8859-1')
#print(df.head())
print(df.shape)
#print(df.info())
#print(df.describe)
print(df.columns)
````

````python
print(df.isnull().sum().sum())
print(df.duplicated().sum())
````

## Define Target Variable

The target variable is `Churn Value`.

It represents the customer's churn status:

- 0 → Customer did not churn
- 1 → Customer churned

Therefore, this is a binary classification problem.

````python
y  = df["Churn Value"]
#print(y.value_counts())
print(y.value_counts(normalize = "index")*100)
````

## Define Feature Matrix

All columns except the target variable are used as input features.

The feature matrix `X` contains the information that the machine learning models will use to predict customer churn.

````python
x = df.drop(["Churn Value"],axis = 1)
print(x.shape)
````

## Identify Feature Types

The feature matrix contains both numerical and categorical variables.

Numerical variables can be directly processed after scaling, while categorical variables need to be converted into numerical form.

Therefore:

- Numerical features → StandardScaler
- Categorical features → OneHotEncoder

````python
numeric_features = x.select_dtypes(include=["int64", "float64"]).columns.tolist()

categorical_features = x.select_dtypes(include=["object"]).columns.tolist()

print("Numerical Features:")
print(numeric_features)

print("\nCategorical Features:")
print(categorical_features)
````

## Train-Test Split

The dataset is divided into training and testing sets.

- 80% of the data will be used for training.
- 20% will be used for testing.

Stratified splitting is used so that the proportion of churned and non-churned customers remains approximately consistent in both datasets.

````python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size = 0.20,random_state= 42,stratify = y)
#check:
print(x_train.shape,x_test.shape)
print(y_train.shape,y_test.shape)
````

## Preprocessing Pipeline

Machine learning algorithms require numerical input.

Numerical features are standardized using StandardScaler.

Categorical features are converted into numerical representations using OneHotEncoder.

A ColumnTransformer is used to apply the appropriate transformation to each feature type.

The preprocessing steps are included inside the machine learning pipeline so that they are fitted only on the training data.

````python
preprocessor = ColumnTransformer(
    transformers = [
        ("numeric",StandardScaler(),numeric_features),
        ("categorical",OneHotEncoder(handle_unknown = "ignore"),categorical_features)
    ]
)
````

### ColumnTransformer — Concept

`ColumnTransformer` is used when **different columns need different preprocessing**.

Example:
- Numerical columns → `StandardScaler`
- Categorical columns → `OneHotEncoder`

### Syntax

```python
ColumnTransformer([
    ("name", transformer, columns)
])

## Model 1: Logistic Regression

Logistic Regression is used as the baseline classification model.

It is a simple and interpretable algorithm for binary classification and provides a useful benchmark for comparing more advanced models.

````python
logistic_model = Pipeline(
    steps = [
        ("Preprocessor",preprocessor),
        ("model",LogisticRegression(max_iter = 1000))                  # max_iter mean maximum iterations
    ]
)
````

````python
logistic_model.fit(x_train,y_train)
````

````python
lr_pred = logistic_model.predict(x_test)
lr_proba = logistic_model.predict_proba(x_test)[:,1]
print(lr_pred[:10])
#print(lr_proba[:10])
````

## Model 2: Decision Tree Classifier

Decision Tree Classifier learns a sequence of decision rules from the data.

It can capture non-linear relationships between customer characteristics and churn.

````python
decision_classifier = Pipeline(
    steps = [
        ("Preprocessor",preprocessor),
        ("model",DecisionTreeClassifier(max_depth = 5,random_state = 42))           # max_depth mean total number of levels for tree
    ]
)
````

````python
decision_classifier.fit(x_train,y_train)
````

````python
dt_pred = decision_classifier.predict(x_test)
print(dt_pred[:10])
````

````python
dt_proba = decision_classifier.predict_proba(x_test)[:,1]
print(dt_proba[:10])
````

## Model 3: Random Forest Classifier

Random Forest is an ensemble learning algorithm that combines multiple decision trees.

It can capture complex relationships in structured customer data and is generally more robust than a single decision tree.

````python
random_forest = Pipeline(
    steps = [
        ("Preprocessor",preprocessor),
        ("model",RandomForestClassifier(n_estimators = 100,random_state = 42))        # n_estimators mean totol number of trees
    ]
)
````

````python
random_forest.fit(x_train,y_train)                            # training of model
````

````python
rf_pred = random_forest.predict(x_test)                        # making predictions
print(rf_pred[:10])
````

````python
rf_proba = random_forest.predict_proba(x_test)[:,1]                # calculating probability
print(rf_proba[:10])
````

## Model 4: XGBoost Classifier

XGBoost is an advanced gradient boosting algorithm.

It builds multiple decision trees sequentially, with each new tree attempting to improve the errors made by previous trees.

It is included to determine whether an advanced boosting approach can provide better predictive performance for customer churn.

- XGBoost doesn't work exactly like Random Forest. Random Forest builds many trees independently and combines their votes, while XGBoost builds trees sequentially, with each new tree trying to improve the previous model.

`eval_metric="logloss" `  This tells XGBoost:

Use Log Loss to evaluate the model's predictions during training.

````python
xgb_classifier = Pipeline(
    steps= [
        ("preprocessor",preprocessor),
        ("model",XGBClassifier(
            n_estimators = 200,
            max_depth= 4,
            learning_rate= 0.05,
            random_state = 42,
            eval_metric = "logloss"

        ))
    ]
)
````

````python
xgb_classifier.fit(x_train,y_train)
````

````python
xgb_pred = xgb_classifier.predict(x_test)
print(xgb_pred[:10])
````

````python
xgb_proba = xgb_classifier.predict_proba(x_test)[:,1]
print(xgb_proba[:10])
````

### Saving Trained Models

`joblib` is a python library and used to **save trained models** so they can be reused later (streamlit) without training them again.

`.pkl` is  basically saved Python object/model file.


````python
import joblib

models = {
    "Logistic Regression": logistic_model,
    "Decision Tree": decision_classifier,
    "Random Forest": random_forest,
    "XGBoost": xgb_classifier
}

joblib.dump(models, "churn_models.pkl")
````

````python
Evaluation_data = {
    "y_test":y_test,
    "x_test":x_test,
    "Predictions":{
        "Logistic Regression": lr_pred,
        "Decision Tree": dt_pred,
        "Random Forest": rf_pred,
        "XGBoost": xgb_pred
    },
    "Probabilities":{
        "Logistic Regression": lr_proba,
        "Decision Tree": dt_proba,
        "Random Forest": rf_proba,
        "XGBoost": xgb_proba
    }
}
joblib.dump(Evaluation_data,"Evaluation_data.pkl")
````
