## Project Overview

Feature Engineering is the process of creating meaningful new features from existing variables.

In customer churn prediction, carefully designed features can help machine learning models identify customer behavior and relationships that may not be obvious from the original variables.

In this notebook, new customer behavior, service usage, tenure, and billing-related features will be created from the cleaned Telco Customer Churn dataset.

The engineered dataset will be used in subsequent exploratory analysis and machine learning stages.

## Objectives

The objectives of this notebook are to:

- Load the cleaned customer churn dataset.
- Examine the available features.
- Create meaningful customer-level features.
- Create service usage features.
- Create tenure-related features.
- Create billing-related features.
- Validate the newly created features.
- Save the engineered dataset for future notebooks.

## Importing Required Libraries:

```python
import pandas as pd
import numpy as np
pd.set_option("display.max_columns", None)
pd.set_option("display.max.rows",None)
```

## Loading Dataset:

```python
file_path = "/content/Cleaned_Customer_Churn.csv"
df = pd.read_csv(file_path,encoding = 'ISO-8859-1')
#print(df.head())
print(df.shape)
print(df.info())
```

## Dataset Check

The cleaned dataset is inspected before feature engineering to confirm that the expected columns and data types are available.

## Total Additional Services

Customers may subscribe to multiple additional services provided by the company.

A new feature, `Total Additional Services`, is created to represent the number of additional services used by each customer.

This feature may help identify whether customers with more services have different churn behavior compared with customers using fewer services.

```python
service_columns = [
    "Online Security",
    "Online Backup",
    "Device Protection",
    "Tech Support",
    "Streaming TV",
    "Streaming Movies",
]
df["Total Aditional Services"] = (df[service_columns]== "Yes").sum(axis = 1)
print(df.head())
```

## Total Streaming Services

The dataset contains two streaming-related services: Streaming TV and Streaming Movies.

A new feature is created to represent the total number of streaming services used by each customer.

```python
streaming_columns = [
    "Streaming TV",
    "Streaming Movies"
]
df["Total Streaming Services"] = (df[streaming_columns] == "Yes").sum(axis=1)
print(df["Total Streaming Services"].value_counts())
```

## Total Security Services

Online Security, Online Backup, and Device Protection represent additional protection-related services.

A combined feature is created to represent the number of security and protection services used by each customer.

```python
security_columns = [
    "Online Security",
    "Online Backup",
    "Device Protection"
    ]
df["Total Security Services"] = (df[security_columns] == "Yes").sum(axis=1)
print(df["Total Security Services"].value_counts().sort_index())
```

## Tenure Group

Customer tenure represents how long a customer has stayed with the company.

To make customer lifecycle stages easier to analyze, customers are grouped into four tenure categories:

- New: 0–12 months
- Early: 13–24 months
- Established: 25–48 months
- Long-term: 49+ months

```python
df["Tenure Group"] = pd.cut(
    df["Tenure Months"],
    [-1, 12, 24, 48,float("inf")],
    labels=["New", "Early", "Established", "Long-term"]
)
print(df["Tenure Group"].value_counts().sort_index())
```

**cut()** is used to convert continuous / numerical variables into meaningful categories.
- `[0,12,24,48,100]` are the boundaries and **labels** are the names for categories.
- `float("inf")` is used for infinite values.

## Multiple Service Indicator

A binary feature identifies customers who use at least two additional services.

This can help investigate whether customers with broader service engagement have different churn behavior.

```python
df["Has Multiple Services"] = (df["Total Aditional Services"] >= 2).astype(int)
print(df["Has Multiple Services"].value_counts())
```
