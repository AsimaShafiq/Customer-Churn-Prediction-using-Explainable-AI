## Import Required Libraries

We import the libraries required to save and load the trained model and preprocessing component.

```python
import joblib
```

## Load the Saved Components

The saved preprocessing component and trained model are loaded so they can be used for prediction without retraining.

```python
preprocessor = joblib.load("/content/Preprocessor.pkl")
model = joblib.load("/content/Model.pkl")
evaluation_data = joblib.load("/content/Evaluation_data.pkl")
print("These components are loaded successfully.")
```

```python
x_test = evaluation_data["x_test"]
y_test = evaluation_data["y_test"]
```

## Verify the Loaded Components

We check the types of the loaded objects to confirm that the correct preprocessing component and model have been loaded.

```python
print(f"preprocessor: {type(preprocessor)}")
print(f"model: {type(model)}")
```

## Prepare New Customer Data

A new customer's information is provided as input to the model.

The input must contain the same features that were used during model training.

We create a DataFrame containing the information of one new customer.

The raw customer information is passed to the saved preprocessor before making a prediction.

```python
input_features = preprocessor.feature_names_in_
print(f"No of input features: {len(input_features)}\n")
print(f"Input features: {input_features}")
```

```python
new_customer = x_test.iloc[[0]].copy()                     # Double square brackets are used to get data in the form of dataframe instead of series.
print(type(new_customer))
print(new_customer.shape)
```

## Apply the Saved Preprocessor

The customer's raw features are transformed using the same preprocessing steps used during model training.

```python
new_customer_processed = preprocessor.transform(new_customer)
print(f"New Customer Processed :{new_customer_processed.shape}")
```

## Generate Churn Prediction

The processed customer data is passed to the trained model to determine whether the customer is predicted to churn.

Class 1 represents Churn and Class 0 represents Non-Churn.

```python
prediction = model.predict(new_customer_processed)[0]                    # predict() returns an array/list for each customerand we are taking 1st element of that array. That's why , we are udsing [0] here.

print(f"Prediction Class: {prediction}")
```

## Generate Churn Probability

The model's probability for the Churn class is calculated to show the customer's estimated churn risk.

```python
probability  = model.predict_proba(new_customer_processed)[0,1]
print(f"Churn Probability: {probability:.2%}")

# predict() → predicted class → [0] = first customer's prediction
# predict_proba() → probabilities for both classes
# [0, 1] → first customer (row 0) + Churn class (column 1)
# Class 0 = Non-Churn | Class 1 = Churn
```

## Compare Actual and Predicted Values

The predicted class is compared with the customer's actual churn label from the test set.

This allows us to determine whether the model classified this test customer correctly.

```python
actual = y_test.iloc[0]
print(f"Actual: {actual}")
print(f"Predicted: {prediction}")
```

```python
if actual == prediction:
    print("Result: Correctly Classified")
else:
    print("Result: Incorrectly Classified")
```

## User-Friendly Prediction Result

The numerical prediction is converted into a simple business-friendly message.

```python
if prediction == 1:
    result = "Likely to Churn"
else:
    result = "Likely to Stay"

print("Prediction:", result)
print(f"Churn Probability: {probability:.2%}")
```

## Create a Reusable Prediction Function

A reusable function is created so that the same prediction process can be applied to any customer with the required input features.
Reusable Churn Prediction Function
This function combines the preprocessing, prediction, and probability calculation steps into a single reusable process. Instead of manually repeating these steps for every new customer, the function can accept any customer data and return a user-friendly churn prediction along with the probability of churn.

```python
def predict_churn(customer_data):
    customer_processed = preprocessor.transform(customer_data)
    prediction = model.predict(customer_processed)[0]
    probability = model.predict_proba(customer_processed)[0, 1]
    if prediction == 1:
        result = "Likely to Churn"
    else:
        result = "Likely to Stay"
    return result, probability
```

## Test the Prediction Function

The reusable prediction function is tested using the selected test customer.

```python
result, probability = predict_churn(new_customer)

print("Prediction:", result)
print(f"Churn Probability: {probability:.2%}")
```
